# Chapter 01 — Exploring FastAPI Basics

Run the server in a separate terminal first:
```bash
uvicorn 01_fastapi_basics.app:app --reload --port 8000
```

Then run the cells below to hit it from Python instead of curl.

````python
import httpx

BASE = "http://localhost:8000"
client = httpx.Client(base_url=BASE, timeout=5.0)
client.get("/").json()
````

````python
# Create a few books
books = [
    {"title": "DDIA", "author": "Kleppmann", "year": 2017},
    {"title": "SICP", "author": "Abelson", "year": 1985},
    {"title": "TPP", "author": "Hunt", "year": 1999},
]
for b in books:
    r = client.post("/books", json=b)
    print(r.status_code, r.json())
````

````python
# Trigger a validation error on purpose — year out of range
r = client.post("/books", json={"title": "", "author": "x", "year": 1})
print(r.status_code)
r.json()
````

````python
client.get("/books", params={"author": "Hunt"}).json()
````
