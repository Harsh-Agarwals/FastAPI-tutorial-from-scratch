# Visualise the request lifecycle

Start the server, then run cells below. Read the response headers and the server logs side by side.

```python
import httpx
r = httpx.get('http://localhost:8000/')
for k, v in r.headers.items():
    print(f'{k}: {v}')
print()
r.json()
```

```python
# Custom request id is echoed back
r = httpx.get('http://localhost:8000/', headers={'X-Request-ID': 'experiment-001'})
r.headers['x-request-id'], r.headers['x-process-time-ms']
```

```python
# /slow endpoint will show ~300ms
httpx.get('http://localhost:8000/slow').headers['x-process-time-ms']
```
