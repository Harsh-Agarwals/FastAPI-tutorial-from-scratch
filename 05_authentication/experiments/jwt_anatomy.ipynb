# JWT anatomy

We mint a token, decode the parts, then verify signature.

```python
import base64, json, httpx
r = httpx.post('http://localhost:8000/token', data={'username': 'ada', 'password': 'password123'})
tok = r.json()['access_token']
header_b64, payload_b64, sig_b64 = tok.split('.')

def b64(x):
    return base64.urlsafe_b64decode(x + '==')

print('HEADER :', json.loads(b64(header_b64)))
print('PAYLOAD:', json.loads(b64(payload_b64)))
print('SIG len:', len(b64(sig_b64)), 'bytes')
```

```python
# Tamper with the payload and see verification fail
from jose import jwt, JWTError
tampered = tok[:-2] + 'AA'
try:
    jwt.decode(tampered, 'dev-secret-do-not-use-in-prod', algorithms=['HS256'])
except JWTError as e:
    print('Rejected as expected:', e)
```
