# Middleware lab

We measure compression savings and timing overhead.

```python
import httpx, time
u = 'http://localhost:8000/big'
raw = httpx.get(u, headers={'Accept-Encoding': 'identity'})
gz = httpx.get(u, headers={'Accept-Encoding': 'gzip'})
print('raw bytes  :', len(raw.content))
print('gzip bytes :', len(gz.content))
print('ratio      :', len(gz.content) / len(raw.content))
```

```python
# Compare timings over N requests
N = 200
u = 'http://localhost:8000/'
client = httpx.Client()
t = time.perf_counter()
for _ in range(N):
    client.get(u)
print(f'avg latency {((time.perf_counter()-t)/N)*1000:.2f} ms — includes middleware overhead')
```
