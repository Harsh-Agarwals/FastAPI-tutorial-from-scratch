# Error envelope tour

Confirm every error path returns the same shape.

```python
import httpx, json
base = 'http://localhost:8000'

calls = [
    ('insufficient', httpx.post, '/transfer', {'json': {'from_account':'acc-1','to_account':'acc-2','amount':9999}}),
    ('not_found',    httpx.post, '/transfer', {'json': {'from_account':'acc-x','to_account':'acc-2','amount':1}}),
    ('validation',   httpx.post, '/transfer', {'json': {'from_account':'','to_account':'','amount':-1}}),
    ('teapot',       httpx.get,  '/teapot',   {}),
    ('boom',         httpx.get,  '/boom',     {}),
]
for name, fn, path, kw in calls:
    r = fn(base + path, **kw)
    print(name, r.status_code, json.dumps(r.json(), indent=2)[:200])
    print('---')
```
