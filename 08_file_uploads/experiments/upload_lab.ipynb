# Upload lab

Programmatic uploads + magic-bytes check.

```python
import httpx, io

fake_pdf = b'%PDF-1.4\n' + b'A' * 1024
real_text = b'hello world\n'

def upload(name, data, ct='application/pdf'):
    files = {'file': (name, io.BytesIO(data), ct)}
    r = httpx.post('http://localhost:8000/upload', files=files)
    return r.status_code, r.json()

print(upload('hello.pdf', fake_pdf))
print(upload('hello.pdf', real_text))  # claims PDF, isn't — 400
print(upload('hello.txt', real_text, ct='text/plain'))
```
