# Polling experiment

Submit a job and poll with exponential backoff.

```python
import httpx, time
r = httpx.post('http://localhost:8000/jobs', json={'text': 'hello' * 50, 'delay_s': 2})
job_id = r.json()['job_id']
print('submitted', job_id)

delay = 0.1
for _ in range(20):
    s = httpx.get(f'http://localhost:8000/jobs/{job_id}').json()
    print(f"{s['status']:8} t={time.strftime('%X')}")
    if s['status'] in ('done', 'failed'):
        print('result:', s.get('result'))
        break
    time.sleep(delay)
    delay = min(delay * 1.6, 1.5)
```
