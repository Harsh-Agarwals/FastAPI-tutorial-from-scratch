# Threading benchmarks

Compare sequential, threaded for IO and CPU.

```python
import time, math
from concurrent.futures import ThreadPoolExecutor

def io_task(ms=100):
    time.sleep(ms/1000)

def cpu_task(n=50_000):
    s = 0
    for i in range(2, n):
        if all(i % p for p in range(2, int(math.isqrt(i))+1)):
            s += i
    return s

def bench(label, fn, n, workers=None):
    t = time.perf_counter()
    if workers is None:
        for _ in range(n): fn()
    else:
        with ThreadPoolExecutor(max_workers=workers) as ex:
            list(ex.map(lambda _: fn(), range(n)))
    print(f'{label:25} workers={workers}  ms={(time.perf_counter()-t)*1000:7.1f}')

for w in (None, 2, 4, 8, 16):
    bench('IO 20x100ms', io_task, 20, w)
print()
for w in (None, 2, 4, 8):
    bench('CPU 8x50k', cpu_task, 8, w)
```
