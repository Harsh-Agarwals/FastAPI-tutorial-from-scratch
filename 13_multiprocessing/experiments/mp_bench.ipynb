# Process-pool benchmarks

Heads-up: in notebooks, `if __name__ == '__main__'` guards still apply for spawn-mode.

```python
import math, time, os
from concurrent.futures import ProcessPoolExecutor, ThreadPoolExecutor

def cpu_task(n=80_000):
    s = 0
    for i in range(2, n):
        if all(i % p for p in range(2, int(math.isqrt(i))+1)):
            s += i
    return s

def bench(label, Pool, n, workers):
    t = time.perf_counter()
    with Pool(max_workers=workers) as ex:
        list(ex.map(cpu_task, [80_000]*n))
    print(f'{label:15} workers={workers}  ms={(time.perf_counter()-t)*1000:7.1f}')

n = (os.cpu_count() or 4)
bench('threads', ThreadPoolExecutor, n, n)
bench('processes', ProcessPoolExecutor, n, n)
```
