# The showdown

Plot all four strategies for each workload.

```python
import httpx, pandas as pd
rows = []
for kind in ('io', 'cpu', 'pdf'):
    r = httpx.get(f'http://localhost:8000/benchmark/{kind}?n=8', timeout=60).json()
    for strat, ms in r['results_ms'].items():
        rows.append({'kind': kind, 'strategy': strat, 'ms': ms})
df = pd.DataFrame(rows)
df.pivot(index='kind', columns='strategy', values='ms')
```

```python
import matplotlib.pyplot as plt
df.pivot(index='kind', columns='strategy', values='ms').plot(kind='bar', figsize=(8,4))
plt.ylabel('ms'); plt.title('Same workload, four strategies'); plt.tight_layout()
```
