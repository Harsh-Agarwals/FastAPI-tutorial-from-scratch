# Programmatic WS client

Uses `websockets` library which ships with uvicorn[standard].

```python
import asyncio, json
from websockets.asyncio.client import connect

async def go():
    async with connect('ws://localhost:8000/ws/chat') as ws:
        await ws.send('Explain the GIL in one sentence.')
        while True:
            msg = json.loads(await ws.recv())
            if msg['event'] == 'end':
                break
            if msg['event'] == 'token':
                print(msg['data'], end='', flush=True)

await go()
```
