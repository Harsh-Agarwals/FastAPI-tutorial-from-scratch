# Consume an SSE stream from Python

`httpx` exposes a streaming API; we parse the SSE protocol by hand.

```python
import httpx, json

with httpx.stream('GET', 'http://localhost:8000/stream/sse?prompt=hi') as r:
    event, data = None, ''
    for line in r.iter_lines():
        if not line:
            if event == 'token':
                print(json.loads(data)['token'], end=' ', flush=True)
            event, data = None, ''
        elif line.startswith('event:'):
            event = line.split(':',1)[1].strip()
        elif line.startswith('data:'):
            data = line.split(':',1)[1].strip()
        if event == 'end':
            break
```

```python
# NDJSON is even simpler
import httpx, json
with httpx.stream('GET', 'http://localhost:8000/stream/ndjson') as r:
    for line in r.iter_lines():
        if line:
            print(json.loads(line)['token'], end=' ', flush=True)
```
