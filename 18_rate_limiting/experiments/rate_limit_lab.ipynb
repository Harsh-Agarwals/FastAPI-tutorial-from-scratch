# Hit the limiter and observe 429s

```python
import httpx, collections
stats = collections.Counter()
for _ in range(30):
    r = httpx.get('http://localhost:8000/slowapi-limited')
    stats[r.status_code] += 1
stats
```

```python
# Token bucket lets bursts through; observe pattern over time
import time, httpx
for _ in range(15):
    r = httpx.get('http://localhost:8000/in-memory')
    print(r.status_code, end=' ')
    time.sleep(0.1)
```
