# Hit/miss tour

Measure cold vs warm latency for each cache.

```python
import httpx, time
for url in ('/memcache', '/redis-cache'):
    t = time.perf_counter(); httpx.get(f'http://localhost:8000{url}?q=cold-{time.time()}'); cold = (time.perf_counter()-t)*1000
    q='warm'
    httpx.get(f'http://localhost:8000{url}?q={q}')
    t = time.perf_counter(); httpx.get(f'http://localhost:8000{url}?q={q}'); warm = (time.perf_counter()-t)*1000
    print(f'{url:14} cold={cold:6.1f} ms  warm={warm:6.1f} ms')
```

```python
# single-flight collapsing
import asyncio, httpx
async def bang():
    async with httpx.AsyncClient() as c:
        return await asyncio.gather(*(c.get('http://localhost:8000/single-flight?q=same') for _ in range(20)))
rs = await bang()
print('OK count:', sum(r.status_code == 200 for r in rs))
```
