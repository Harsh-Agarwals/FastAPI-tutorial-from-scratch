# Talk to Redis directly

Skip FastAPI — just exercise the data types.

```python
import redis.asyncio as ar
r = ar.from_url('redis://localhost:6379/0', decode_responses=True)
await r.set('hi', 'there', ex=10)
print(await r.get('hi'))
for u, s in [('alice', 50), ('bob', 80), ('carol', 65)]:
    await r.zadd('lb', {u: s})
print(await r.zrevrange('lb', 0, -1, withscores=True))
await r.close()
```
