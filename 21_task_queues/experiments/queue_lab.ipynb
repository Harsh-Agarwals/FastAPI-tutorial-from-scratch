# Drive the queue

Make sure the worker process is running.

```python
import httpx, time
ids = []
for i in range(8):
    r = httpx.post('http://localhost:8000/jobs', json={'text': f'job {i}'})
    ids.append(r.json()['job_id'])
print(ids)
```

```python
import time, httpx
for _ in range(20):
    depth = httpx.get('http://localhost:8000/queue/depth').json()
    states = [httpx.get(f'http://localhost:8000/jobs/{j}').json().get('status') for j in ids]
    print(depth, states)
    if all(s in ('done','failed') for s in states): break
    time.sleep(1)
```
