# Use the LLM client without the HTTP layer

Great for testing prompt changes without server round-trips.

```python
import sys, pathlib, asyncio
sys.path.append(str(pathlib.Path.cwd().parent))
from app import get_llm, build_prompt, call_with_retry

llm = get_llm()
prompt = build_prompt('The quick brown fox jumps over the lazy dog.', style='bullet')
res = await call_with_retry(llm, prompt)
res
```

```python
# Show retries actually firing — call the mock many times
import statistics, time
times = []
for _ in range(20):
    t = time.perf_counter()
    try:
        await call_with_retry(llm, prompt)
    except Exception as e:
        print('exhausted retries:', e)
    times.append((time.perf_counter()-t)*1000)
print('p50 ms:', statistics.median(times), ' p95 ms:', sorted(times)[int(0.95*len(times))])
```
