# Visualise the toy embeddings

We project the 16-d toy vectors down to 2-d with PCA.

```python
import sys, pathlib
sys.path.append(str(pathlib.Path.cwd().parent))
from app import mock_embed, CORPUS
import numpy as np
X = np.stack([mock_embed(t) for t in CORPUS])
X.shape
```

```python
# Simple PCA to 2-d
Xc = X - X.mean(0)
U, S, V = np.linalg.svd(Xc, full_matrices=False)
P = Xc @ V[:2].T

import matplotlib.pyplot as plt
plt.figure(figsize=(7,5))
for (x, y), t in zip(P, CORPUS):
    plt.scatter(x, y); plt.text(x+0.005, y, t[:30], fontsize=8)
plt.title('Toy embeddings (PCA→2D)')
```

```python
# Pairwise cosine
from itertools import combinations
for i, j in combinations(range(len(CORPUS)), 2):
    print(f'{np.dot(X[i], X[j]):+.3f}  {CORPUS[i][:30]!r:35s} <-> {CORPUS[j][:30]!r}')
```
