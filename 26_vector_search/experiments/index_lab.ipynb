# Drive the vector index

```python
import httpx
items = [
    {'text': 'FastAPI is great for APIs.', 'tags': ['python','web']},
    {'text': 'Async code does not block.', 'tags': ['python','async']},
    {'text': 'Pasta with tomato sauce.', 'tags': ['food']},
    {'text': 'The event loop juggles coroutines.', 'tags': ['async']},
]
print(httpx.post('http://localhost:8000/index/items', json={'items': items}).json())
```

```python
for q in ('event loop', 'food', 'rest api'):
    print('Q:', q)
    for r in httpx.get('http://localhost:8000/search', params={'q': q, 'k': 3}).json():
        print(f"  {r['score']:+.3f}  {r['text']}")
    print()
```
