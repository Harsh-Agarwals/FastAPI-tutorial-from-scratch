# PDF chunking lab

Generate a synthetic PDF with reportlab... or skip and feed plain text into the chunker.

```python
import sys, pathlib
sys.path.append(str(pathlib.Path.cwd().parent))
from app import normalise, chunk_text

pages = [
    'FastAPI is an async-first web framework for Python.  '*40,
    'It builds on Starlette and Pydantic for high performance.  '*40,
    'In this experiment we chunk a tiny corpus and inspect the boundaries.  '*40,
]
chunks = chunk_text(pages, target_chars=600, overlap_chars=80)
for c in chunks:
    print(f'p{c.page_start}-{c.page_end}  [{c.char_start}:{c.char_end}]  {c.text[:80]}…')
```

```python
# Confirm overlap exists
for a, b in zip(chunks, chunks[1:]):
    print('overlap len:', a.char_end - b.char_start)
```
