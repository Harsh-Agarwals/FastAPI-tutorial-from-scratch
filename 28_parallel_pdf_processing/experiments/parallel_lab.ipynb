# Parallel summarisation lab

Drive the endpoint with a real PDF.

```python
import httpx, pathlib
pdf_path = pathlib.Path('sample.pdf')   # put one next to this notebook
if pdf_path.exists():
    files = {'file': (pdf_path.name, pdf_path.read_bytes(), 'application/pdf')}
    r = httpx.post('http://localhost:8000/summarize-pdf', files=files, timeout=120)
    print(r.status_code)
    print('chunks:', r.json().get('chunks'))
    print('summary:', r.json().get('summary')[:500])
else:
    print('Drop a sample.pdf in this folder to run.')
```
