# See workers in action

Start uvicorn with `--workers 4` and run this notebook.

```python
import httpx, collections
pids = collections.Counter()
for _ in range(200):
    pids[httpx.get('http://localhost:8000/who').json()['pid']] += 1
pids
```

```python
# Counter is monotonic across all workers (Redis-backed)
for _ in range(20):
    print(httpx.post('http://localhost:8000/count').json()['count'], end=' ')
```
