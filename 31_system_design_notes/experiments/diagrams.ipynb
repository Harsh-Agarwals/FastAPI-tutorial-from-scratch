```python
# Tiny capacity calculator — tweak the numbers
DOCS_PER_MIN  = 100
CHUNKS        = 12
LLM_LATENCY_S = 0.7
CONCURRENCY   = 5
REDUCE_S      = 0.5

wall = CHUNKS / CONCURRENCY * LLM_LATENCY_S + REDUCE_S
rps  = DOCS_PER_MIN / 60 * (CHUNKS + 1)
workers_per_box = max(1, round(DOCS_PER_MIN / 60 * wall * 2))
print(f'wall per doc  : {wall:.2f} s')
print(f'LLM RPS needed: {rps:.1f}')
print(f'~workers/box   : {workers_per_box}')
```
