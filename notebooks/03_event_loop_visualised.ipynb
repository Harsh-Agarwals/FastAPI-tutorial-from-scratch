# Event loop visualised

We log each step a set of coroutines takes. The print order is the order the event loop ran them.

```python
import asyncio, time
T0 = time.perf_counter()
def stamp(): return f'{(time.perf_counter()-T0)*1000:6.1f}ms'

async def worker(name, steps):
    for s in steps:
        print(stamp(), name, 'before', s)
        await asyncio.sleep(s)
        print(stamp(), name, 'after ', s)

await asyncio.gather(
    worker('A', [0.1, 0.05, 0.1]),
    worker('B', [0.07, 0.07, 0.07]),
    worker('C', [0.05, 0.2, 0.05]),
)
```

## Observation

Interleaved output proves coroutines yield at every `await`. The event loop juggled three of them without any threads.
