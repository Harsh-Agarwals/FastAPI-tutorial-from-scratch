# Cache hit ratio under different access patterns

Even a tiny LRU is dramatically helpful on **Zipfian** traffic.

```python
import random, collections

def zipf_sample(n_keys, alpha=1.2, count=10000):
    weights = [1/(i+1)**alpha for i in range(n_keys)]
    return random.choices(range(n_keys), weights=weights, k=count)

def lru_hit_ratio(keys, cap):
    cache = collections.OrderedDict()
    hits = misses = 0
    for k in keys:
        if k in cache:
            cache.move_to_end(k); hits += 1
        else:
            cache[k] = True; misses += 1
            if len(cache) > cap: cache.popitem(last=False)
    return hits / (hits + misses)

for pattern, keys in [
    ('uniform 1000 keys', [random.randrange(1000) for _ in range(10000)]),
    ('zipf 1000 keys',    zipf_sample(1000)),
    ('zipf 100k keys',    zipf_sample(100_000)),
]:
    for cap in (10, 100, 1000):
        print(f'{pattern:20s} cap={cap:5d} hit={lru_hit_ratio(keys, cap):.2%}')
```
