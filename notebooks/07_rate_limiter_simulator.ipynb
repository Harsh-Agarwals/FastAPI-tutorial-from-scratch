# Token bucket vs sliding window

Simulate a burst and see who gets through.

```python
import collections, time, random

def token_bucket_sim(events, capacity, refill_per_sec):
    tokens = capacity
    last = 0.0
    accepted = rejected = 0
    for t in events:
        tokens = min(capacity, tokens + (t - last) * refill_per_sec)
        last = t
        if tokens >= 1:
            tokens -= 1; accepted += 1
        else:
            rejected += 1
    return accepted, rejected

def sliding_window_sim(events, window_s, limit):
    q = collections.deque()
    accepted = rejected = 0
    for t in events:
        while q and q[0] <= t - window_s:
            q.popleft()
        if len(q) < limit:
            q.append(t); accepted += 1
        else:
            rejected += 1
    return accepted, rejected

# Burst of 100 events in 1 second, then steady at 5 rps for 9 seconds.
events = sorted([random.uniform(0, 1.0) for _ in range(100)] + [1 + i*0.2 for i in range(45)])
print('token_bucket cap=20 refill=10:', token_bucket_sim(events, 20, 10))
print('sliding 10s limit=60        :', sliding_window_sim(events, 10, 60))
```
